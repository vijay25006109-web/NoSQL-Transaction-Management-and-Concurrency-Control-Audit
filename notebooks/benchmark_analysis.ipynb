# MongoDB benchmark analysis
Run the Python script with `--benchmark`, enter the measured values in the CSV, then calculate and compare latency/throughput. This notebook intentionally contains no fabricated measurements.

```
import pandas as pd
from pathlib import Path
p=Path('../reports/benchmark_calculation_sheet.csv')
df=pd.read_csv(p)
df
```

```
measured=df.dropna(subset=['elapsed_ms']).copy()
if not measured.empty:
    measured['calculated_avg_ms']=measured.elapsed_ms/measured.operations
    measured['calculated_ops_per_second']=measured.operations/(measured.elapsed_ms/1000)
measured
```

Interpretation: compare read latency and throughput alongside documents/keys examined. Consider index write overhead and repeat tests on identical conditions.
